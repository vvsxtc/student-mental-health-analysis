```python
students = {
    "Student_ID": [1, 2, 3, 4],
    "Age": [19, 20, 21, 19],
    "Sleep_Hours": [7, 5, 8, 6],
    "Stress_Level": ["Low", "High", "Low", "Medium"],
    "Study_Hours": [3, 5, 2, 4],
    "Academic_Score": [85, 72, 90, 78]
}

print(students)
```

```python
import pandas as pd

df = pd.DataFrame(students)

df
```

```python
import matplotlib.pyplot as plt

df.groupby("Stress Level")["Academic Score"].mean().plot(kind="bar")

plt.title("Average Academic Score by Stress Level")
plt.xlabel("Stress Level")
plt.ylabel("Average Academic Score")

plt.show()
```
